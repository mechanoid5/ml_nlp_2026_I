```python
from keys import OPENROUTER_API_KEY
```

```python
MODEL_ID = "nvidia/nemotron-3-nano-30b-a3b:free"
# MODEL_ID = "qwen/qwen3-4b:free"
```

```python
# messages=[ {"role": "user", "content": "How many r's are in the word 'strawberry'?" } ]
messages=[ {"role": "user", "content": "Объясни принцип работы нейронных сетей простыми словами. Ответ должен быть кратким до 100 слов."} ]
```

```python
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=OPENROUTER_API_KEY,
)
```

```python
%%time

response = client.chat.completions.create(
  model=MODEL_ID,
  messages=messages,
  extra_body={"reasoning": {"enabled": False}}
)
```
